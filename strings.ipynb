```python
# Q26 — Mini Business Case 🔥

# An online store has:

# visitors = 25000
# orders = 850
# revenue = 1275000
# marketing_cost = 180000

# Calculate:

# Conversion rate %
# AOV
# Marketing cost per order
# Revenue after marketing cost
# Whether the store generated more than ₹1 million after marketing cost

visitors = 25000
orders = 850
revenue = 1275000
marketing_cost = 180000
cvr=(orders/visitors)*100
aov= revenue/orders
profit=revenue-marketing_cost
print(f'cvr: {cvr:.2}%\nAOV: {aov} profit: {profit}\n Whether the store generated more than ₹1 million after marketing cost: {bool(profit>1000000)} ')
```

```python
orders = [500, 1200, 7500, 3000, 15000, 800]
category= []
for i in orders:
    if i<1000:
        category.append('Low')
    elif 1000<=i<5000:
        category.append('Medium')
    else:
        category.append('High')
category
```

```python
orders = [500, 1200, 7500, 3000, 15000, 800]

classification = [
    "Low" if order < 1000
    else "Medium" if order < 5000
    else "High"
    for order in orders
]

print(classification)
```

```python
customers = {
    "C001": {"name": "Shorya", "revenue": 120000},
    "C002": {"name": "Rahul", "revenue": 45000},
    "C003": {"name": "Amit", "revenue": 250000},
    "C004": {"name": "Priya", "revenue": 80000}
}
segments = {}

for customer_id, customer in customers.items():
    if customer["revenue"] >= 100000:
        segments[customer_id] = "High Value"
    elif customer["revenue"] >= 50000:
        segments[customer_id] = "Medium Value"
    else:
        segments[customer_id] = "Low Value"

segments
```

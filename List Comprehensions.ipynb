# List Comprehensions

```python
numbers = [1, 2, 3, 4, 5]
squares= [n**2 for n in numbers]
squares
```

```python
numbers = [1, 2, 3, 4, 5]
```

```python
cubes=[n**3 for n in numbers]
cubes
```

```python
numbers = [10, 15, 20, 25, 30]
updated=[n for n in numbers if n>20]
updated
```

```python
numbers = [1, 2, 3, 4, 5, 6]
odd=[n for n in numbers if n%2!=0]
odd
```

```python
products = ["laptop", "mouse", "keyboard", "monitor"]
upper= [prod.upper() for prod in products]
upper
```

```python
sales = [500, 1200, 800, 2500, 900, 3000]
high_sales=[n for n in sales if n>=1000]
high_sales
```

```python
prices = [100, 200, 300, 400]
hike= [n+n*(0.1) for n in prices]
hike
```

```python
names = ["shorya", "rahul", "amit"]
caps= [n.title() for n in names]
caps
```

```python
numbers = [1, 2, 3, 4, 5]
odd=[n for n in numbers if n%2 !=0]
even=[n for n in numbers if n%2 ==0]
print(even, odd)
```

```python
orders = [1200, 450, 2300, 800, 1750, 3000, 650]
high_hike=[n+(n*0.05) for n in orders if n>1000]
high_hike
```

Q10 — Real Business Task 🔥

Given:

revenue = [120000, 85000, 150000, 95000, 200000]

Calculate:

total revenue
average revenue
highest revenue
lowest revenue
difference between highest and lowest revenue

Then determine whether the average revenue is greater than ₹1,00,000 using a Boolean expression.

```python
revenue = [120000, 85000, 150000, 95000, 200000]
total_rev=sum(revenue)
avg_rev=total_rev/len(revenue)
max_rev=max(revenue)
low_rev=min(revenue)
range_rev=max_rev-low_rev
print(f'total_rev: {total_rev}\navg_rev: {avg_rev}\nHighest Revenue: {max_rev}\nLowest Revenue: {low_rev}\ndifference between highest and lowest revenue: {range_rev}')
print(f'whether the average revenue is greater than ₹1,00,000: {bool(avg_rev>100000)}')
```
